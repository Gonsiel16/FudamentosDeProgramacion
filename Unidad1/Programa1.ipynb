# Tema de listas


```python
alumnos = ["Ana", "Luis", "Carlos", "Juan"]
print(alumnos)
print(alumnos[0])
print(alumnos[3])
print(alumnos[5]) #marca error por que el indice o posicion no existe
```

```python
#Modificar un valor de la lista
alumnos[1]= "Maria Luisa"
print(alumnos)
```

```python
#El metodo append agrega un elemento al final de la lista
alumnos.append("Karla")
print(alumnos)
```

```python
#El metodo insert agrega un elemento a la lista en la posición especificada
alumnos.insert(2, "Cristina")
print(alumnos)
```

```python
#remove elimina un elemento especificando el valor del elemento
#debe coincidir exactamente mayusculas y minusculas
alumnos.remove("Ana")
alumnos.remove("ana") #este marcaría error por que no esta bien escrito el valor
print(alumnos)
```

```python
#pop elimina el elemento o item en la posición especificada
alumnos.pop(2)
print(alumnos)
```

```python
#del tambien elimina el elemento o item en la posición especificada
del alumnos[3]
print(alumnos)
```

```python
#La función len me permite saber cuantos elementos tiene una lista
print("elementos en la lista de alumnos:", len(alumnos))
```

```python
# add Ana and Ernesto and show the list
alumnos.append("Ana")
alumnos.append("Ernesto")
for alumno in alumnos:
    print(alumno)
```

```python
print(alumnos)
for alumno in alumnos:
    print(alumno)
    if alumno == "Ana":
        break #break forza a salir del ciclo siempre y cuando se cumpla la funcion del if
```
