```python
# ========================================
# SISTEMA DE PRESTAMOS
# Version 2 - Listas, historial y estadisticas
# ========================================

from datetime import datetime

# Lista de alumnos: numero de control, nombre y carrera
alumnos = [
    ["26040200", "Superman", "Ingenieria en Inteligencia Artificial"],
    ["26040201", "Bat Man", "Ingenieria en Inteligencia Artificial"],
    ["26040202", "Wonder Woman", "Ingenieria Informatica"],
    ["26040203", "Flash", "Ingenieria Informatica"],
    ["26040204", "Supergirl", "Ingenieria en Inteligencia Artificial"]
]

computadoras = ["PC-LC1-01", "PC-LC1-02", "PC-LC1-03",
                "PC-LC2-01", "PC-LC2-02", "PC-LC2-03",
                "PC-LC3-01", "PC-LC3-02", "PC-LC3-03", "PC-LC3-04"]

aulas = ["LC1", "LC1", "LC1",
         "LC2", "LC2", "LC2",
         "LC3", "LC3", "LC3", "LC3"]

software = ["Python, VS Code", "Python, Java, VS Code", "AutoCAD, MATLAB",
            "Python, VS Code", "Python, Java, VS Code", "AutoCAD, MATLAB",
            "Python, VS Code", "Python, Java, VS Code", "AutoCAD, MATLAB", "Python, VS Code"]

estados = ["Disponible", "Prestada", "Reservada",
           "Disponible", "Prestada", "Disponible",
           "Disponible", "Prestada", "Reservada", "Mantenimiento"]

# Historiales
prestamos = []       # [numero_control, programa, aula, id_computadora, carrera]
necesidades = []     # [programa, fecha_hora]

# MENU PRINCIPAL
opcion = 0
while opcion != 10:
    print("\n================================")
    print("      SISTEMA DE PRESTAMOS")
    print("================================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar prestamo")
    print("4. Devolver computadora")
    print("5. Crear reservacion (pendiente)")
    print("6. Agregar computadora")
    print("7. Historial de prestamos")
    print("8. Necesidades de software")
    print("9. Estadisticas")
    print("10. Salir")
    print("================================")

    try:
        opcion = int(input("Seleccione una opcion: "))
    except ValueError:
        print("Debe ingresar un numero.")
        continue

    # 1. Consultar computadoras
    if opcion == 1:
        print("\n--- COMPUTADORAS ---")
        for i in range(len(computadoras)):
            print("Aula:", aulas[i], "| ID:", computadoras[i],
                  "| Software:", software[i], "| Estado:", estados[i])

    # 2. Buscar por software
    elif opcion == 2:
        programa = input("\nIngrese el software que desea buscar: ").strip()
        encontrado = False
        print("\n--- RESULTADOS ---")
        for i in range(len(computadoras)):
            if programa.lower() in software[i].lower():
                print("ID:", computadoras[i])
                print("Aula:", aulas[i])
                print("Software:", software[i])
                print("Estado:", estados[i])
                encontrado = True
        if not encontrado:
            print("No se encontraron computadoras con ese software.")

    # 3. Solicitar prestamo
    elif opcion == 3:
        print("\n=== SOLICITAR PRESTAMO ===")
        num_control = input("Numero de control del alumno: ").strip()
        indice_alumno = -1

        # Buscar numero de control en la primera columna de alumnos
        for i in range(len(alumnos)):
            if alumnos[i][0] == num_control:
                indice_alumno = i
                break

        if indice_alumno == -1:
            print("El alumno no existe en la lista.")
            continue

        nombre = alumnos[indice_alumno][1]
        carrera = alumnos[indice_alumno][2]
        programa = input("Que programa necesita?: ").strip()

        # Revisar si alguna computadora tiene ese programa
        hay_programa = False
        asignado = False

        for i in range(len(computadoras)):
            if programa.lower() in [p.strip().lower() for p in software[i].split(",")]:
                hay_programa = True
                if estados[i] == "Disponible":
                    estados[i] = "Prestada"
                    prestamos.append([num_control, programa, aulas[i],
                                      computadoras[i], carrera])
                    print("Prestamo registrado para", nombre)
                    print("Se asigno la computadora", computadoras[i],
                          "del aula", aulas[i])
                    asignado = True
                    break

        if not hay_programa:
            respuesta = input("Ese programa no esta registrado. Desea agregarlo a la lista de necesidades? (s/n): ").strip().lower()
            if respuesta == "s":
                fecha_hora = datetime.now().strftime("%d/%m/%Y %H:%M:%S")
                necesidades.append([programa, fecha_hora])
                print("Programa agregado a la lista de necesidades.")
            else:
                print("No se agrego el programa.")
        elif not asignado:
            print("El software existe, pero no hay computadoras disponibles con ese programa.")

    # 4. Devolver computadora
    elif opcion == 4:
        print("\n=== DEVOLVER COMPUTADORA ===")
        id_pc = input("Dame el ID de la computadora: ").strip()
        existe = False

        for i in range(len(computadoras)):
            if computadoras[i] == id_pc:
                existe = True
                if estados[i] == "Prestada":
                    estados[i] = "Disponible"
                    print("La computadora", id_pc, "ya esta disponible.")
                else:
                    print("La computadora no esta prestada; su estado es:", estados[i])
                break

        if not existe:
                     print("No existe la computadora", id_pc)

    # 5. Crear reservacion (todavia no implementada)
    elif opcion == 5:
        print("\nLa opcion de crear reservacion aun no esta implementada.")

    # 6. Agregar computadora
    elif opcion == 6:
        print("\n--- AGREGAR COMPUTADORA ---")
        id_computadora = input("ID de la computadora: ").strip()
        if id_computadora in computadoras:
            print("Ya existe una computadora con ese ID.")
        else:
            aula = input("Aula: ").strip()
            programas = input("Software instalado (separado por comas): ").strip()
            computadoras.append(id_computadora)
            aulas.append(aula)
            software.append(programas)
            estados.append("Disponible")
            print("Computadora agregada correctamente.")

    # 7. Historial de prestamos
    elif opcion == 7:
        print("\n--- HISTORIAL DE PRESTAMOS DE COMPUTADORAS ---")
        if len(prestamos) == 0:
            print("Todavia no hay prestamos registrados.")
        else:
            for i in range(len(prestamos)):
                print(i + 1, "| Alumno:", prestamos[i][0],
                      "| Software:", prestamos[i][1],
                      "| Aula:", prestamos[i][2],
                      "| ID:", prestamos[i][3])

    # 8. Necesidades de software
    elif opcion == 8:
        print("\n--- HISTORIAL DE NECESIDADES ---")
        if len(necesidades) == 0:
            print("No hay necesidades de software registradas.")
        else:
            for i in range(len(necesidades)):
                print(i + 1, "| Programa:", necesidades[i][0],
                      "| Fecha:", necesidades[i][1])

    # 9. Estadisticas
    elif opcion == 9:
        print("\n======= ESTADISTICAS POR PROGRAMA =======")
        conteo_programas = {}
        conteo_carreras = {}

        for prestamo in prestamos:
            programa = prestamo[1].strip().lower()
            carrera = prestamo[4]

            if programa in conteo_programas:
                conteo_programas[programa] += 1
            else:
                conteo_programas[programa] = 1

            if carrera in conteo_carreras:
                conteo_carreras[carrera] += 1
            else:
                conteo_carreras[carrera] = 1

        if len(prestamos) == 0:
            print("Todavia no hay prestamos para calcular estadisticas.")
        else:
            for programa in conteo_programas:
                print(programa, ":", conteo_programas[programa], "veces")

            print("\n======= ESTADISTICAS POR CARRERA =======")
            for carrera in conteo_carreras:
                print(carrera + ":", conteo_carreras[carrera], "veces")
```
